GENERIC WAREHOUSE INVENTORY SYSTEM

TRIAL 2: WAREHOUSE INVENTORY SYSTEM

```python
from typing import Generic, TypeVar
T = TypeVar("T")

class Inventory(Generic[T]):
    def __init__(self):
        self.items = []

    def store(self, item: T):
        self.items.append(item)

    def retrieve(self):
        return self.items

    def display(self):
        print("Inventory Items:")
        for item in self.items:
            print(f'Type:{type(item).__name__}, Value: {item}')
```

PRODUCT NAMES (STRINGS)

```python
inventory1 = Inventory[str]()
inventory1.store("Laptop")
inventory1.store("Gamepad")
inventory1.store("Monitor")

print("\nInventories:")
inventory1.display()
```

STOCK QUANTITIES (INTEGERS)

```python
stock = Inventory[int]()
stock.store(21)
stock.store(45)

print("\nStocks:")
stock.display()
```

PRODUCT PRICES (FLOATS/LISTS)

```python
serial_number = Inventory[list]()
serial_number.store(["FN014","CE419","EL079"])
serial_number.store(["RN112","LC123","RB045"])

print("\nSerial Numbers:")
serial_number.display()
```
